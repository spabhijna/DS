2. Queue Using two Stacks

```python
def two_stacks(max_sum,a,b):
  count = 0
  current_sum,i,j = 0,0,0

  while i<len(a) and current_sum + a[i] <= max_sum:
    current_sum += a[i]
    i += 1
    count += 1

  max_count = count


  while j < len(b):
    current_sum += b[j]
    j += 1
    while current_sum > max_sum:
      i -=1
      current_sum -= a[i]
    if current_sum <= max_sum:
      count += 1
      max_count = max(max_count,i+j)

  return max_count
```

```python
a = [4,3,4,6,1]
b = [2,1,8,5]
max_sum = 10
print(two_stacks(max_sum,a,b))
```
